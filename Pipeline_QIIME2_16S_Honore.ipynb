# **Ne choisissez pas encore les paramètres DADA2 au hasard**

Les valeurs de trunc-len-f et trunc-len-r devront être déterminées à partir du graphique de qualité. Elles doivent aussi conserver un chevauchement suffisant :trunc-len-f + trunc-len-r − longueur de l’amplicon

**Lecture du graphique**

Les lectures **Forward** restent de bonne qualité jusqu’à environ **240–250 bases**. À la position **243**, la médiane est encore à **Q30** et le premier quartile à **Q25**.

Les lectures **Reverse** se dégradent plus rapidement, surtout après environ **190–210 bases**. À la position **247**, la médiane n’est plus que de **Q24** et le premier quartile de **Q1**5. Il est donc déconseillé de conserver les Reverse jusqu’à **247–250 bases**.

**Paramètres recommandés**

Pour ce jeu de données d’apprentissage, utilisez: trunc-len-f = 250
trunc-len-r = 200 *texte en italiqueQIIME 2 View_HT22.07.2026.jpg*

Le notebook prévoit déjà le retrait de 19 bases Forward et 20 bases Reverse, vraisemblablement pour retirer les amorces. Après traitement, les longueurs utiles seront donc approximativement :
Forward : 250 - 19 = 231 bases
Reverse : 200 - 20 = 180 bases **texte en gras**

# **Quel réglage utiliser pour trunc-len-f et trunc-len-r?**

````python
!qiime dada2 denoise-paired \
  --i-demultiplexed-seqs demux-paired-end.qza \
  --o-table table-emp \
  --o-representative-sequences rep-seqs \
  --o-denoising-stats denoising-stats.qza \
  --p-n-threads 1 \
  --p-trim-left-f 19 \
  --p-trim-left-r 20 \
  --p-trunc-len-f 250 \
  --p-trunc-len-r 200
````

````python
Pourquoi selon vous le pourcentage de séquence retenue à la fin est-il si faible?
(Comparez avec le fichier denoising-stats.qzv avec toutes les séquences dans output_16S): Le faible pourcentage de séquences retenues est principalement attribuable au sous-échantillonnage des données, limité à environ 2 000 lectures par échantillon. DADA2 apprend son modèle d’erreurs à partir des séquences disponibles et utilise leur abondance pour distinguer les variantes biologiques réelles des erreurs. Avec peu de lectures, le modèle est moins robuste et les variants rares sont insuffisamment représentés, ce qui entraîne l’élimination d’une plus grande proportion de séquences. L’analyse du jeu complet dans output_16S fournit davantage d’information pour l’apprentissage des erreurs et conduit donc à un meilleur taux de rétention.
````

# **Regrouper tous vos outputs dans un seul dossier**

Exécutez cette cellule :

````python
from pathlib import Path
import shutil

project_dir = Path("/content/AtelierSYMBIOSOL")
output_dir = project_dir / "output_Honore"

output_dir.mkdir(parents=True, exist_ok=True)

# Copier les principaux formats de résultats présents
extensions = [
    "*.qza",
    "*.qzv",
    "*.tsv",
    "*.csv",
    "*.biom",
    "*.fasta",
    "*.log"
]

for pattern in extensions:
    for source_file in project_dir.glob(pattern):
        shutil.copy2(source_file, output_dir / source_file.name)

# Copier les résultats exportés, s'ils existent
exported_folders = [
    "denoising-stats-exported"
]

for folder_name in exported_folders:
    source_folder = project_dir / folder_name
    destination_folder = output_dir / folder_name

    if source_folder.exists():
        shutil.copytree(
            source_folder,
            destination_folder,
            dirs_exist_ok=True
        )

# Copier les métadonnées
metadata_file = project_dir / "data" / "metadata.tsv"

if metadata_file.exists():
    shutil.copy2(
        metadata_file,
        output_dir / "metadata.tsv"
    )

print("Outputs regroupés dans :", output_dir)
````
