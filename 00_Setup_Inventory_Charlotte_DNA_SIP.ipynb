Étape 8 — Inventorier tous les FASTQ

```python
from pathlib import Path
import pandas as pd
import re

fastq_files = sorted(
    list(RAW_ROOT.rglob("*.fastq.gz"))
    + list(RAW_ROOT.rglob("*.fq.gz"))
    + list(RAW_ROOT.rglob("*.fastq"))
    + list(RAW_ROOT.rglob("*.fq"))
)

records = []

for file_path in fastq_files:
    filename = file_path.name

    if re.search(r"(^|[_\-.])R1([_\-.]|$)", filename, re.IGNORECASE):
        direction = "R1"
    elif re.search(r"(^|[_\-.])R2([_\-.]|$)", filename, re.IGNORECASE):
        direction = "R2"
    else:
        direction = "Non déterminée"

    relative_path = file_path.relative_to(RAW_ROOT)
    path_parts = relative_path.parts

    run = path_parts[0] if len(path_parts) >= 1 else "Inconnu"
    marker = path_parts[1] if len(path_parts) >= 2 else "Inconnu"

    records.append({
        "run": run,
        "marker": marker,
        "direction": direction,
        "filename": filename,
        "size_MB": round(file_path.stat().st_size / 1024**2, 2),
        "absolute_path": str(file_path.resolve()),
    })

inventory = pd.DataFrame(records)

if inventory.empty:
    print("Aucun FASTQ trouvé.")
else:
    display(inventory)
    print("\nNombre total de fichiers :", len(inventory))
    display(
        inventory.groupby(
            ["run", "marker", "direction"],
            dropna=False
        ).size().reset_index(name="nombre_fichiers")
    )
```

Enregistrez l’inventaire :

```python
inventory_file = LOG_ROOT / "fastq-inventory.tsv"

inventory.to_csv(
    inventory_file,
    sep="\t",
    index=False
)

print("Inventaire enregistré dans :", inventory_file)
```

**Étape 9 — Vérifier les couples R1/R2**

```python
import pandas as pd

pair_records = []

r1_files = sorted(
    list(RAW_ROOT.rglob("*_R1_*.fastq.gz"))
    + list(RAW_ROOT.rglob("*_R1_*.fq.gz"))
)

for r1 in r1_files:
    r2_name = r1.name.replace("_R1_", "_R2_", 1)
    r2 = r1.parent / r2_name

    provisional_sample_id = re.sub(
        r"_S\d+_L\d+_R1_\d+\.(fastq|fq)\.gz$",
        "",
        r1.name,
        flags=re.IGNORECASE
    )

    relative_path = r1.relative_to(RAW_ROOT)
    path_parts = relative_path.parts

    pair_records.append({
        "run": path_parts[0],
        "marker": path_parts[1],
        "sample_id_provisional": provisional_sample_id,
        "R1": r1.name,
        "R2": r2.name if r2.exists() else "",
        "pair_complete": r2.exists(),
        "R1_path": str(r1.resolve()),
        "R2_path": str(r2.resolve()) if r2.exists() else "",
    })

pairs = pd.DataFrame(pair_records)

display(pairs)

if not pairs.empty:
    print(
        "Couples complets :",
        int(pairs["pair_complete"].sum()),
        "/",
        len(pairs)
    )

    incomplete = pairs.loc[~pairs["pair_complete"]]

    if not incomplete.empty:
        print("\nAttention : fichiers R2 absents")
        display(incomplete)
```

Enregistrez la table :

```python
pairs_file = LOG_ROOT / "fastq-pairs.tsv"

pairs.to_csv(
    pairs_file,
    sep="\t",
    index=False
)

print("Table des couples enregistrée dans :", pairs_file)
```
